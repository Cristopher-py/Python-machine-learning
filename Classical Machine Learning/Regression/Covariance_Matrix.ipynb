## Covariance matrix of OLS parameter estimates

We assume the true data-generating process $x_i - y_i$ is given by:

$$
    \mathbf{y_{\text{true}}} = \mathcal{X} \boldsymbol{\beta} + \boldsymbol{\varepsilon}
$$

Where:
- $\mathbf{y_{\text{true}}}$ is the vector of observed responses,
- $\mathcal{X}$ is the design matrix (full column rank),
- $\boldsymbol{\beta}$ is the vector of true parameters,
- $\boldsymbol{\varepsilon}$ is the vector of random errors.

We assume the errors satisfy the classical spherical errors conditions:

\begin{cases}
    \mathbb{E}[\boldsymbol{\varepsilon}] = \mathbf{0} = \left( 0, 0, 0, \dots 0 \right)^{T} \\
    \operatorname{Var}(\boldsymbol{\varepsilon}) = \sigma^{2} \, \mathbf{1} = \sigma^{2} \, \left( 1, 1, 1, \dots 1 \right)^{T} \\
    \operatorname{Cov}(\boldsymbol{\varepsilon}) = \sigma^{2} \, \mathcal{I}
\end{cases}

Here $\sigma$ is an unknown real and positive constant, and $\mathcal{I}$ denotes the identity matrix (its dimension depends on context).

Below we expand each expression componentwise to clarify its meaning:

\begin{cases}
    \mathbb{E}[\varepsilon_i] = 0 \quad \text{for all } i\\
    \operatorname{Var}(\varepsilon_i) = \sigma^{2} \quad \text{for all } i\\
    \operatorname{Cov}(\varepsilon_i, \, \varepsilon_j) = 0 \quad \text{for all } i \neq j
\end{cases}

The OLS estimator $\hat{\boldsymbol{\beta}}$ minimizes the sum of squared residuals and is obtained from the OLS normal equations:

$$
    \mathcal{X}^{T} \mathcal{X} \, \hat{\boldsymbol{\beta}} = \mathcal{X}^{T} \mathbf{y_{\text{true}}}
$$

If $\mathcal{X}^{T} \mathcal{X}$ is invertible (no perfect multicollinearity), we have:

$$
    \hat{\boldsymbol{\beta}} = (\mathcal{X}^{T} \mathcal{X})^{-1} \mathcal{X}^{T} \mathbf{y_{\text{true}}}
$$

Substitute the true model $\mathbf{y_{\text{true}}} = \mathcal{X}\boldsymbol{\beta} + \boldsymbol{\varepsilon}$ into the OLS formula:

\begin{aligned}
    \hat{\boldsymbol{\beta}}
    &= (\mathcal{X}^{T} \mathcal{X})^{-1} \mathcal{X}^{T} (\mathcal{X}\boldsymbol{\beta} + \boldsymbol{\varepsilon}) \\
    &= \boldsymbol{\beta} + (\mathcal{X}^{T} \mathcal{X})^{-1} \mathcal{X}^{T} \boldsymbol{\varepsilon}
\end{aligned}

Hence, the estimation error is:

$$
    \hat{\boldsymbol{\beta}} - \boldsymbol{\beta} = (\mathcal{X}^{T} \mathcal{X})^{-1} \mathcal{X}^{T} \boldsymbol{\varepsilon}
$$

We can easily see that $\hat{\boldsymbol{\beta}}$ is unbiased, that means:

$$
    \mathbb{E} [\hat{\boldsymbol{\beta}}] = \boldsymbol{\beta}
$$

Due to the zero mean of $\boldsymbol{\varepsilon}$.

By definition, the covariance matrix of $\hat{\boldsymbol{\beta}}$ is:

$$
    \operatorname{Cov}(\hat{\boldsymbol{\beta}}) = \mathbb{E}\!\left[ (\hat{\boldsymbol{\beta}} - \boldsymbol{\beta})(\hat{\boldsymbol{\beta}} - \boldsymbol{\beta})^T \right]
$$

Insert the error expression. Because $(\mathcal{X}^{T} \mathcal{X})^{-1}$ is symmetric:

\begin{aligned}
    \operatorname{Cov}(\hat{\boldsymbol{\beta}})
    &= \mathbb{E}\!\left[ (\mathcal{X}^{T} \mathcal{X})^{-1} \mathcal{X}^{T} \boldsymbol{\varepsilon} \;
       \bigl( (\mathcal{X}^{T} \mathcal{X})^{-1} \mathcal{X}^{T} \boldsymbol{\varepsilon} \bigr)^T \right] \\
    &= (\mathcal{X}^{T} \mathcal{X})^{-1} \mathcal{X}^{T} \,
       \mathbb{E}\!\left[ \boldsymbol{\varepsilon} \, \boldsymbol{\varepsilon}^{T} \right] \,
       \mathcal{X} (\mathcal{X}^{T} \mathcal{X})^{-1}
\end{aligned}

Now use the spherical errors property $\mathbb{E}[\boldsymbol{\varepsilon} \, \boldsymbol{\varepsilon}^{T}] = \sigma^{2} \, \mathcal{I}$:

\begin{aligned}
    \operatorname{Cov}(\hat{\boldsymbol{\beta}})
    &= (\mathcal{X}^{T} \mathcal{X})^{-1} \mathcal{X}^{T} (\sigma^{2} \, \mathcal{I}) \mathcal{X} (\mathcal{X}^{T} \mathcal{X})^{-1} \\
    &= \sigma^{2} \, (\mathcal{X}^{T} \mathcal{X})^{-1} \mathcal{X}^{T} \mathcal{X} (\mathcal{X}^{T} \mathcal{X})^{-1}
\end{aligned}

Since $\mathcal{X}^{T} \mathcal{X} \, (\mathcal{X}^{T} \mathcal{X})^{-1} = \mathcal{I}$, we obtain the important result:

$$
    \boxed{\operatorname{Cov}(\hat{\boldsymbol{\beta}}) = \sigma^{2} \, (\mathcal{X}^{T} \mathcal{X})^{-1}}
$$

We don't know $\sigma$ yet. To estimate it, we can use the residuals. The vector of fitted values is $\hat{\mathbf{y}} = \mathcal{X} \hat{\boldsymbol{\beta}}$, so the residuals are:

$$
    \mathbf{r} = \mathbf{y_{\text{true}}} - \hat{\mathbf{y}} = \mathbf{y_{\text{true}}} - \mathcal{X} \hat{\boldsymbol{\beta}}
$$

Substitute $\mathbf{y_{\text{true}}} = \mathcal{X}\boldsymbol{\beta} + \boldsymbol{\varepsilon}$ and $\hat{\boldsymbol{\beta}} = \boldsymbol{\beta} + (\mathcal{X}^T \mathcal{X})^{-1} \mathcal{X}^T \boldsymbol{\varepsilon}$:

\begin{aligned}
    \mathbf{r} &= \mathcal{X}\boldsymbol{\beta} + \boldsymbol{\varepsilon} - \mathcal{X}\bigl[ \boldsymbol{\beta} + (\mathcal{X}^T \mathcal{X})^{-1} \mathcal{X}^T \boldsymbol{\varepsilon} \bigr] \\
    &= \boldsymbol{\varepsilon} - \mathcal{X} (\mathcal{X}^T \mathcal{X})^{-1} \mathcal{X}^T \boldsymbol{\varepsilon}
\end{aligned}

Define the projection matrix $\mathcal{P} = \mathcal{X} (\mathcal{X}^T \mathcal{X})^{-1} \mathcal{X}^T$, which is symmetric $\mathcal{P}^{T} = \mathcal{P}$ and idempotent $\mathcal{P}^2 = \mathcal{P}$. Same properties for $\mathcal{I} - \mathcal{P}$ Then:

$$
    \mathbf{r} = (\mathcal{I} - \mathcal{P}) \, \boldsymbol{\varepsilon}
$$

The minimized sum of squared residuals is:

$$
    \text{SSR} = \mathbf{r}^T \mathbf{r} = \boldsymbol{\varepsilon}^T (\mathcal{I} - \mathcal{P})^T (\mathcal{I} - \mathcal{P}) \, \boldsymbol{\varepsilon}
$$

Because $\mathcal{I} - \mathcal{P}$ is symmetric and idempotent, we have:

$$
    \text{SSR} = \boldsymbol{\varepsilon}^T (\mathcal{I} - \mathcal{P}) \, \boldsymbol{\varepsilon}
$$

This is just a dot product:

$$
    \text{SSR} = \sum_{i} \, \varepsilon_{i}^{2} - \sum_{i} \, p_{ii} \, \varepsilon_{i}^{2} - 2 \, \sum_{i < j} \, p_{ij} \, \varepsilon_{i} \, \varepsilon_{j}
$$

Applying the linear operator $\mathbb{E}$, accounting to spherical errors, we get:

$$
    \mathbb{E}[\text{SSR}] = n \, \sigma^{2} - \sum_{i} \, \mathbb{E} \left[ p_{ii} \, \varepsilon_{i}^{2} \right]
$$

Where $n$ is the total number of data observed. This term $\sum_{i} \, \mathbb{E} \left[ p_{ii} \, \varepsilon_{i}^{2} \right]$ is simply:

$$
    \sum_{i} \, \mathbb{E} \left[ p_{ii} \, \varepsilon_{i}^{2} \right] = \sigma^{2} \, \sum_{i} \, p_{ii} = k \, \sigma^{2}
$$

Where $k$ is the total number of model's parameters. Therefore:

$$
    \mathbb{E}[\text{SSR}] = (n - k) \, \sigma^{2}
$$

From this expression, we obtain an unbiased estimator of $\sigma^{2}$:

$$
    \boxed{\hat{\sigma}^{2} = \frac{\text{SSR}}{n - k}}
$$

In fact:

$$
    \mathbb{E}\left[\hat{\sigma}^{2} \right] = \frac{\mathbb{E}[\text{SSR}]}{n - k} = \sigma^{2}
$$

## Covariance matrix of WLS parameter estimates

We now assume that the errors are uncorrelated but heteroscedastic, with known variances (in the Ordinary Least Squares scenario, we assumed spherical errors). Specifically, we know the variance of each $\varepsilon_i$ up to a common scale factor $\sigma^{2}$:

\begin{cases}
    \mathbb{E}[\boldsymbol{\varepsilon}] = \mathbf{0} \\
    \operatorname{Var}(\boldsymbol{\varepsilon}) = \sigma^{2} \, \mathcal{W}^{-1}
\end{cases}

Here $\sigma$ is an unknown positive constant, and $\mathcal{W} = \operatorname{diag}(w_1, w_2, \dots, w_n)$ is a known diagonal matrix of positive weights. The relation $\operatorname{Var}(\varepsilon_i) = \sigma^{2} / w_i$ holds for each $i$. The weight matrix $\mathcal{W}$ is symmetric and positive definite.

Below we expand each expression componentwise to clarify its meaning:

\begin{aligned}
    &\mathbb{E}[\varepsilon_i] = 0 \quad \text{for all } i \\
    &\operatorname{Var}(\varepsilon_i) = \sigma^{2} / w_i \quad \text{for all } i \\
    &\operatorname{Cov}(\varepsilon_i, \, \varepsilon_j) = 0 \quad \text{for all } i \neq j
\end{aligned}

The Weighted Least Squares estimator $\hat{\boldsymbol{\beta}}$ minimizes the weighted sum of squared residuals:

$$
    \hat{\boldsymbol{\beta}} = \arg\min_{\boldsymbol{\beta}} \; (\mathbf{y_{\text{true}}} - \mathcal{X}\boldsymbol{\beta})^{T} \, \mathcal{W} \, (\mathbf{y_{\text{true}}} - \mathcal{X}\boldsymbol{\beta})
$$

Differentiating with respect to $\boldsymbol{\beta}$ yields the WLS normal equations:

$$
    \mathcal{X}^{T} \mathcal{W} \, \mathcal{X} \, \hat{\boldsymbol{\beta}} = \mathcal{X}^{T} \mathcal{W} \, \mathbf{y_{\text{true}}}
$$

If $\mathcal{X}^{T} \mathcal{W} \, \mathcal{X}$ is invertible (no perfect multicollinearity), we have:

$$
    \hat{\boldsymbol{\beta}} = (\mathcal{X}^{T} \mathcal{W} \, \mathcal{X})^{-1} \mathcal{X}^{T} \mathcal{W} \, \mathbf{y_{\text{true}}}
$$

Substitute the true model $\mathbf{y_{\text{true}}} = \mathcal{X}\boldsymbol{\beta} + \boldsymbol{\varepsilon}$ into the WLS formula:

\begin{aligned}
    \hat{\boldsymbol{\beta}}
    &= (\mathcal{X}^{T} \mathcal{W} \, \mathcal{X})^{-1} \mathcal{X}^{T} \mathcal{W} \, (\mathcal{X}\boldsymbol{\beta} + \boldsymbol{\varepsilon}) \\
    &= \boldsymbol{\beta} + (\mathcal{X}^{T} \mathcal{W} \, \mathcal{X})^{-1} \mathcal{X}^{T} \mathcal{W} \, \boldsymbol{\varepsilon}
\end{aligned}

Hence, the estimation error is:

$$
    \hat{\boldsymbol{\beta}} - \boldsymbol{\beta} = (\mathcal{X}^{T} \mathcal{W} \, \mathcal{X})^{-1} \mathcal{X}^{T} \mathcal{W} \, \boldsymbol{\varepsilon}
$$

Because $\mathbb{E}[\boldsymbol{\varepsilon}] = \mathbf{0}$, we see that $\hat{\boldsymbol{\beta}}$ is unbiased:

$$
    \mathbb{E}[\hat{\boldsymbol{\beta}}] = \boldsymbol{\beta}
$$

By definition, the covariance matrix of $\hat{\boldsymbol{\beta}}$ is:

$$
    \operatorname{Cov}(\hat{\boldsymbol{\beta}}) = \mathbb{E}\!\left[ (\hat{\boldsymbol{\beta}} - \boldsymbol{\beta})(\hat{\boldsymbol{\beta}} - \boldsymbol{\beta})^T \right]
$$

Insert the error expression. Since $(\mathcal{X}^{T} \mathcal{W} \, \mathcal{X})^{-1}$ is symmetric:

\begin{aligned}
    \operatorname{Cov}(\hat{\boldsymbol{\beta}})
    &= \mathbb{E}\!\left[ (\mathcal{X}^{T} \mathcal{W} \, \mathcal{X})^{-1} \mathcal{X}^{T} \mathcal{W} \, \boldsymbol{\varepsilon} \;
       \bigl( (\mathcal{X}^{T} \mathcal{W} \, \mathcal{X})^{-1} \mathcal{X}^{T} \mathcal{W} \, \boldsymbol{\varepsilon} \bigr)^T \right] \\
    &= (\mathcal{X}^{T} \mathcal{W} \, \mathcal{X})^{-1} \mathcal{X}^{T} \mathcal{W} \,
       \mathbb{E}\!\left[ \boldsymbol{\varepsilon} \, \boldsymbol{\varepsilon}^{T} \right] \,
       \mathcal{W} \, \mathcal{X} (\mathcal{X}^{T} \mathcal{W} \, \mathcal{X})^{-1}
\end{aligned}

Now use the heteroscedasticity property $\mathbb{E}[\boldsymbol{\varepsilon} \, \boldsymbol{\varepsilon}^{T}] = \sigma^{2} \, \mathcal{W}^{-1}$:

\begin{aligned}
    \operatorname{Cov}(\hat{\boldsymbol{\beta}})
    &= (\mathcal{X}^{T} \mathcal{W} \, \mathcal{X})^{-1} \mathcal{X}^{T} \mathcal{W} (\sigma^{2} \, \mathcal{W}^{-1}) \mathcal{W} \, \mathcal{X} (\mathcal{X}^{T} \mathcal{W} \, \mathcal{X})^{-1} \\
    &= \sigma^{2} \, (\mathcal{X}^{T} \mathcal{W} \, \mathcal{X})^{-1} \mathcal{X}^{T} \mathcal{W} \, \mathcal{W}^{-1} \mathcal{W} \, \mathcal{X} (\mathcal{X}^{T} \mathcal{W} \, \mathcal{X})^{-1}
\end{aligned}

Because $\mathcal{W} \, \mathcal{W}^{-1} = \mathcal{I}$, this simplifies to:

$$
    \operatorname{Cov}(\hat{\boldsymbol{\beta}}) = \sigma^{2} \, (\mathcal{X}^{T} \mathcal{W} \, \mathcal{X})^{-1} \mathcal{X}^{T} \mathcal{W} \, \mathcal{X} (\mathcal{X}^{T} \mathcal{W} \, \mathcal{X})^{-1}
$$

Since $\mathcal{X}^{T} \mathcal{W} \, \mathcal{X} \, (\mathcal{X}^{T} \mathcal{W} \, \mathcal{X})^{-1} = \mathcal{I}$, we obtain the important result:

$$
    \boxed{\operatorname{Cov}(\hat{\boldsymbol{\beta}}) = \sigma^{2} \, (\mathcal{X}^{T} \mathcal{W} \, \mathcal{X})^{-1}}
$$

We don't know $\sigma$ yet. To estimate it, we use the weighted residuals. The vector of fitted values is $\hat{\mathbf{y}} = \mathcal{X} \hat{\boldsymbol{\beta}}$, so the residuals are:

$$
    \mathbf{r} = \mathbf{y_{\text{true}}} - \hat{\mathbf{y}} = \mathbf{y_{\text{true}}} - \mathcal{X} \hat{\boldsymbol{\beta}}
$$

Substitute $\mathbf{y_{\text{true}}} = \mathcal{X}\boldsymbol{\beta} + \boldsymbol{\varepsilon}$ and $\hat{\boldsymbol{\beta}} = \boldsymbol{\beta} + (\mathcal{X}^T \, \mathcal{W} \, \mathcal{X})^{-1} \mathcal{X}^T \, \mathcal{W} \, \boldsymbol{\varepsilon}$:

\begin{aligned}
    \mathbf{r} &= \mathcal{X}\boldsymbol{\beta} + \boldsymbol{\varepsilon} - \mathcal{X}\bigl[ \boldsymbol{\beta} + (\mathcal{X}^T \, \mathcal{W} \, \mathcal{X})^{-1} \mathcal{X}^T \, \mathcal{W} \, \boldsymbol{\varepsilon} \bigr] \\
    &= \boldsymbol{\varepsilon} - \mathcal{X} (\mathcal{X}^T \, \mathcal{W} \, \mathcal{X})^{-1} \mathcal{X}^T \, \mathcal{W} \, \boldsymbol{\varepsilon}
\end{aligned}

Define the WLS projection matrix $\mathcal{P}_{\mathcal{W}} = \mathcal{X} (\mathcal{X}^T \, \mathcal{W} \, \mathcal{X})^{-1} \mathcal{X}^T \, \mathcal{W}$. It satisfies $\mathcal{P}_{\mathcal{W}}^2 = \mathcal{P}_{\mathcal{W}}$ (idempotent) and $\mathcal{P}_{\mathcal{W}} \, \mathcal{X} = \mathcal{X}$. The complementary matrix $\mathcal{I} - \mathcal{P}_{\mathcal{W}}$ has the property $(\mathcal{I} - \mathcal{P}_{\mathcal{W}}) \mathcal{X} = \mathbf{0}$, so:

$$
    \mathbf{r} = (\mathcal{I} - \mathcal{P}_{\mathcal{W}}) \, \boldsymbol{\varepsilon}
$$

The weighted sum of squared residuals is:

$$
    \text{SSR} = \mathbf{r}^T \mathcal{W} \mathbf{r} = \boldsymbol{\varepsilon}^T (\mathcal{I} - \mathcal{P}_{\mathcal{W}})^T \mathcal{W} (\mathcal{I} - \mathcal{P}_{\mathcal{W}}) \, \boldsymbol{\varepsilon}
$$

Using the fact that $\mathcal{P}_{\mathcal{W}}^T \mathcal{W} = \mathcal{W} \mathcal{P}_{\mathcal{W}}$ (which follows from symmetry of $\mathcal{W}$ and the definition of $\mathcal{P}_{\mathcal{W}}$), one can show:

$$
    (\mathcal{I} - \mathcal{P}_{\mathcal{W}})^T \mathcal{W} (\mathcal{I} - \mathcal{P}_{\mathcal{W}}) = \mathcal{W} - \mathcal{W} \mathcal{P}_{\mathcal{W}}
$$

Thus:

$$
    \text{SSR} = \boldsymbol{\varepsilon}^T (\mathcal{W} - \mathcal{W} \mathcal{P}_{\mathcal{W}}) \, \boldsymbol{\varepsilon}
$$

Now take the expectation. Because $\mathbb{E}[\boldsymbol{\varepsilon}] = \mathbf{0}$ and $\operatorname{Var}(\boldsymbol{\varepsilon}) = \sigma^{2} \mathcal{W}^{-1}$, we have for any matrix $\mathbf{A}$: $\mathbb{E}[\boldsymbol{\varepsilon}^T \mathbf{A} \boldsymbol{\varepsilon}] = \operatorname{tr}(\mathbf{A} \operatorname{Var}(\boldsymbol{\varepsilon}))$. Hence:

\begin{aligned}
    \mathbb{E}[\text{SSR}] &= \operatorname{tr}\bigl( (\mathcal{W} - \mathcal{W} \mathcal{P}_{\mathcal{W}}) \, \sigma^{2} \mathcal{W}^{-1} \bigr) \\
    &= \sigma^{2} \, \operatorname{tr}\bigl( \mathbf{I} - \mathcal{P}_{\mathcal{W}} \bigr)
\end{aligned}

The trace of $\mathcal{P}_{\mathcal{W}}$ equals the rank of $\mathcal{X}$, which is $k$ (the number of parameters). Therefore:

$$
    \mathbb{E}[\text{SSR}] = \sigma^{2} (n - k)
$$

From this expression, we obtain an unbiased estimator of $\sigma^{2}$:

$$
    \boxed{\hat{\sigma}^{2} = \frac{\text{SSR}}{n - k}}
$$

Indeed:

$$
    \mathbb{E}\!\left[\hat{\sigma}^{2} \right] = \frac{\mathbb{E}[\text{SSR}]}{n - k} = \sigma^{2}
$$

This is the general framework.

### If we know the variance of $y_{\text{true}}$

If we know the variance of each $y_{i}$, we can set the weights $w_{i}$ equal to:

$$
    \boxed{w_{i} = \frac{1}{\sigma_{i}^{2}}}
$$

Then we can absorb $\sigma$ by setting $\sigma = 1$. In this case, we have:

$$
    \boxed{\operatorname{Cov}(\hat{\boldsymbol{\beta}}) = (\mathcal{X}^{T} \mathcal{W} \, \mathcal{X})^{-1}}
$$

And we don't need to estimate $\sigma$ anymore.
